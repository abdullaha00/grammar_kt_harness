# One record through the current pipeline

This notebook calls the same stage functions as `scripts/run.py`. Fixture-backed model responses keep it cost-free and deterministic.

```python
import sys
from pathlib import Path
ROOT = Path.cwd()
sys.path.insert(0, str(ROOT / 'src'))
from grammar_kt.io import load_experiment, load_typed_resource
from grammar_kt.normalise import normalise
from grammar_kt.canonicalise import canonicalise
from grammar_kt.generate import generate_items
from grammar_kt.validate_items import validate_items
from grammar_kt.fold import apply_fold
from grammar_kt.simulate import simulate
```

```python
config = load_experiment('fixture')
resources = load_typed_resource(config['resource']['data'], config['resource']['schema'])
mappings = normalise(resources, config['normalisation'])
cells = canonicalise(mappings, config['canonical']['schema'])
candidates = generate_items(cells, config['generation'])
items, judgments = validate_items(candidates, cells, config['validation'])
grammar_fold = apply_fold(cells, config['fold'])
events = simulate(items, grammar_fold, config['simulation'])
```

```python
print('typed EGP input:', resources[0])
print('normalised mapping:', mappings[0])
print('GrammarCell:', cells[0])
print('accepted item:', items[0])
print('first BaseEvent:', events[0])
```
